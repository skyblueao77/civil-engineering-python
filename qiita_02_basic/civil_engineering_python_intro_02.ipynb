## 1. 変数
変数とは、データを保存しておくための箱のようなものです。

> **補足**  
> 初心者向けのプログラミングの一般的な説明では「変数はデータを保存する箱」と表現されることがあります。  
> ただし、Pythonでは厳密には変数がオブジェクトを参照する仕組みになっています。

例えば次のコードを見てみましょう。


```python
a = [1, 2, 3]
b = a
b[0] = 4
print(a)
```

`b` のデータを変更したにもかかわらず、`a` の内容も変化しています。  
これは `a` と `b` が同じリストオブジェクトを参照しているためです。  
そのため、Pythonの変数は「データに名前を付けるもの」や「オブジェクトへの参照」と考える方がより正確です。

例えば橋の長さを保存したい場合は次のように書きます。


```python
bridge_length = 120
print(bridge_length)
```

`=` は数学のイコールではなく、**右側の値を左側に代入する**という意味です。

## 2. データ型
Pythonではデータの種類ごとに型が存在します。代表的なものを紹介します。

| 型      | 例       | 説明   |
|:--------|:---------|:-------|
| `int`   | `100`    | 整数   |
| `float` | `12.5`   | 小数   |
| `str`   | `"Kobe"` | 文字列 |
| `bool`  | `True`   | 真偽値 |

### 整数 (int)


```python
traffic_count = 1500
print(traffic_count)
```

### 小数 (float)

```python
river_width = 25.6
print(river_width)
```

### 文字列 (str)

```python
city_name = "Kobe"
print(city_name)
```

### 真偽値 (bool)

```python
is_safe = True
print(is_safe)
```

## 3. 四則演算
Pythonでは電卓のような計算ができます。


```python
a = 10
b = 5
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

土木と絡めると例えば道路延長の計算などにも利用できます。

```python
road1 = 120
road2 = 80
total = road1 + road2
print(total)
```

## 4. 条件分岐
条件によって処理を変える場合は `if` を使います。  
例えば降雨量を判定してみます。


```python
rainfall = 120
if rainfall > 100:
    print("大雨に注意")
```

### else
条件に当てはまらない場合は `else` を使います。


```python
rainfall = 50
if rainfall > 100:
    print("大雨に注意")
else:
    print("通常")
```

## 5. 繰り返し処理
同じ処理を何度も行う場合は `for` を使います。


```python
for i in range(5):
    print(i)
```

### 土木っぽい例
5地点の測定データを順番に表示してみます。


```python
depth = [1.2, 1.5, 1.8, 2.0, 1.7]
for d in depth:
    print(d)
```

## 6. リスト
複数のデータをまとめて扱う場合はリストを使います。


```python
population = [1000, 1200, 1500]
print(population)
```

交通量や人口データなどを扱う際によく利用されます。

## 7. 簡単な演習
3地点の交通量データがあるとします。合計値を計算してみましょう。


```python
traffic = [1200, 1500, 1800]
total = 0
for t in traffic:
    total += t
print(total)
```

このように複数地点のデータをまとめて計算できます。
