## 1. pandasとは

pandasは、表形式のデータを効率よく扱うためのPythonライブラリです。

Excelのような表データをPythonから読み込んだり、検索したり、集計したりすることができます。

例えば、土木分野では次のようなデータを扱うことがあります。

- 水門の開度や流量
- 河川の水位
- 雨量
- 交通量
- 地盤調査結果
- 人口統計
- 気象観測データ

これらのデータは、多くの場合「行」と「列」を持つ表形式になっています。

例えば、次のような水門データを考えてみます。

| 水門 | 開度(m) | 流量(m3/s) |
| -- | ----- | -------- |
| A  | 1.2   | 32       |
| B  | 2.0   | 55       |
| C  | 1.5   | 40       |

このようなデータについて、

- 平均流量を求める
- 最大流量を調べる
- 流量が一定値を超える水門を探す
- 必要な列だけを取り出す
- CSVファイルから大量のデータを読み込む

といった処理を、pandasを使うことで簡単に行えるようになります。

pandasを使うメリットは、単に大量のデータを扱えることだけではありません。条件による抽出、集計、並べ替え、欠測値の処理、データの結合など、データ分析で頻繁に行う処理を比較的少ないコードで実行できます。(近年`Polars`などのライブラリも使われるようになってきていますが、現時点ではpandasが最も普及しているため、本シリーズではpandasを扱います。)

## 3. DataFrameを作ってみる

まずは簡単な水門データを作成してみます。

````python
import pandas as pd

gate_data = pd.DataFrame(
    {
        "水門": ["A", "B", "C"],
        "開度(m)": [1.2, 2.0, 1.5],
        "流量(m3/s)": [32, 55, 40]
    }
)

print(gate_data)
````

この表のようなデータ構造を **DataFrame** と呼びます。

DataFrameには列名（カラム名）があり、今回の例では

- 水門
- 開度(m)
- 流量(m3/s)

が列名にあたります。

### DataFrameとSeries

pandasでは、表全体を **DataFrame** と呼びます。

一方、DataFrameから1つの列を**Series**として取り出すことができます。
イメージ図↓
```txt
DataFrame（表全体）

| 水門 | 開度(m) | 流量(m3/s) |
|------|---------|-----------|
| A    | 1.2     | 32        |
| B    | 2.0     | 55        |
| C    | 1.5     | 40        |

↓ 流量列を取り出す

Series

0    32
1    55
2    40
```
このように、DataFrameは複数のSeriesから構成されています。

例えば、

````python
print(gate_data["流量(m3/s)"])
````

とすると、「流量(m3/s)」という1列だけを取り出すことができます。

このような1次元のデータが`Series`です。

イメージとしては、

- DataFrame → 表全体
- Series → 表の1列

と考えておくと分かりやすいでしょう。

pandasでは、このDataFrameとSeriesを使ってデータを操作していきます。

## 4. 平均流量を求める

流量の平均値を計算してみます。

````python
print(gate_data["流量(m3/s)"].mean())
````

平均流量は約42.3m³/sであることが分かります。

----

## 5. 最大流量を求める

最大値も簡単に計算できます。

````python
print(gate_data["流量(m3/s)"].max())
````

最大流量は55m³/sです。

----

## 6. 条件に合うデータを探す

流量が40m³/sを超える水門だけを表示してみます。

````python
result = gate_data[gate_data["流量(m3/s)"] > 40]

print(result)
````

このように条件に基づいてデータを抽出できます。

### 必要な列だけを取り出す

例えば、「流量が40m³/sを超える水門について、水門名と流量だけを表示したい」とします。

````python
result = gate_data.loc[
    gate_data["流量(m3/s)"] > 40,
    ["水門", "流量(m3/s)"]
]

print(result)
````

locを使うと、条件に合う行だけでなく、表示したい列も指定できます。
locの基本形は

```python
df.loc[行条件, 列名]
```

です。

左側で**行**を指定し、右側で**列**を指定します。
このように、pandasでは必要なデータだけを取り出して分析することができます。

## 9. データをグループごとに集計する

pandasには、同じ種類のデータをグループに分けて集計する機能もあります。

例えば、次のような交通量データがあるとします。

````python
traffic_data = pd.DataFrame(
    {
        "曜日": ["月", "月", "火", "火", "水", "水"],
        "地点": ["A", "B", "A", "B", "A", "B"],
        "交通量": [1200, 1500, 1300, 1600, 1250, 1550]
    }
)

print(traffic_data)
````

曜日ごとの平均交通量を求める場合は、次のようにします。

````python
print(traffic_data.groupby("曜日")["交通量"].mean())
````

`groupby()` は「同じ値を持つ行をまとめて集計する機能」と考えるとイメージしやすいでしょう。

土木分野でも、

- 地点ごとの平均交通量
- 日ごとの平均雨量
- 観測地点ごとの平均水位

など、グループごとの集計を行う場面があります。

今回は基本的な使い方だけ紹介しましたが、`groupby()` はpandasでデータ分析を行う上で重要な機能です。
